```python
import pandas as pd

df = pd.read_csv("BikeData.csv")
df
```

1. Создать Total_count (сумма Partner 1 и Partner 2)

```python
df['Total_count'] = df['Partner 1'] + df['Partner 2']

df[['Partner 1', 'Partner 2', 'Total_count']]
```

2. Удалить колонки Partner 1 и Partner 2

```python
df = df.drop(['Partner 1', 'Partner 2'], axis=1)
df
```

3. Преобразовать переменную Date в datetime. dayfirst=True - парсит даты в формате день-месяц-год, rrors='coerce' - если отказано, то NaN

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')
df
```

4. Сколько велосипедов выдано в день/месяц, число которого совпадает с днём выполнения задания. Какой это день недели?

4.1. Добавляем дополнительные колонки

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Weekday'] = df['Date'].dt.day_name()

df
```

4.2. Текущая дата. Timestamp почти то же что datetime.datetime

```python
today = pd.Timestamp.now().date()
today_day = today.day
today_month = today.month
```

4.3. Фильтруем данные, где совпадают день и месяц

```python
same_day_data = df.loc[
    (df['Day'] == today_day) & (df['Month'] == today_month),
    ['Date', 'Total_count', 'Weekday']
].copy()
```

4.4. Группируем по году, чтобы увидеть аренды за разные годы в этот день

```python
bikes_by_year = df.loc[
    (df['Day'] == today_day) & (df['Month'] == today_month)
].groupby('Year')['Total_count'].sum().reset_index()
```

4.5. Определяем день недели сегодняшней даты

```python
weekday_today = pd.Timestamp.now().day_name()
```

4.6. Вывод результатов

```python
print(f"Сегодня: {today_day:02d}.{today_month:02d}")
print(f"День недели: {weekday_today}")
print("\nКоличество аренд в этот день по разным годам:")
display(bikes_by_year)

print("\nПодробные данные за этот день:")
display(same_day_data)
```

5. Перекодировать категориальную переменную Functioning Day ("Yes" = True, "No" = False) и посчитать количество нерабочих дней (дней, в которых работало < 12 часов)

```python
# Yes -> True, No -> False
df['Functioning Day'] = df['Functioning Day'].apply(lambda x: True if x == 'Yes' else False)

# Считаем количество часов работы пункта в день
hours_period_day = df.groupby(df['Date'].dt.date)['Functioning Day'].sum().reset_index(name='Hours_open')
display(hours_period_day)

# День считается нерабочим, если работали < 12 часов
hours_period_day['Non_working_day'] = hours_period_day['Hours_open'] < 12
display(hours_period_day)
# Количество нерабочих дней
num_nonworking_days = hours_period_day['Non_working_day'].sum()

print(f"Количество нерабочих дней: {num_nonworking_days}")
```

6. Перекодировать Holiday в 0/1 (используя lambda)

```python
df['Holiday'] = df['Holiday'].apply(lambda x: 1 if x == 'Holiday' else 0)
df[['Date', 'Holiday']]
```

7. Ввести новую категориальную переменную Temperature category ("Freezing", если температура < 0;
"Chilly", если 0 <= температура < 15;
"Nice", если 15 <= температура < 26;
"Hot", если 26 <= температура.)

```python
def temp_category(temp):
    if temp < 0:
        return "Freezing"
    elif 0 <= temp < 15:
        return "Chilly"
    elif 15 <= temp < 26:
        return "Nice"
    else:
        return "Hot"

df['Temperature category'] = df['Temperature'].apply(temp_category)
df[['Temperature', 'Temperature category']]
```

8. Создать переменную Good weather (1 = хорошая погода) по заданным условиям (Temperature category = "Nice",
Humidity в диапазоне от 40 до 60 ,
скорость ветра Wind speed меньше 5.4,
нет дождя (Rainfall)
нет снега (Snowfall)).

```python
# Создаем новую колонку Good weather
def good_weather(row):
    if (
        row['Temperature category'] == 'Nice' and
        40 <= row['Humidity'] <= 60 and
        row['Wind speed'] < 5.4 and
        row['Rainfall'] == 0 and
        row['Snowfall'] == 0
    ):
        return 1
    else:
        return 0

df['Good weather'] = df.apply(good_weather, axis=1)

# Считаем процент наблюдений с хорошей погодой
percent_good = df['Good weather'].mean() * 100
print(f"Процент хорошей погоды: {percent_good:.2f}%")

df[['Date', 'Temperature', 'Humidity', 'Wind speed', 'Rainfall', 'Snowfall', 'Temperature category', 'Good weather']].head(10)
```

9. Сгруппировать данные по сезонам и категории температуры

```python
# Через groupby
group = df.groupby(['Seasons', 'Temperature category'])['Total_count'].sum().reset_index()

# Через pivot_table
pivot = df.pivot_table(index='Seasons', columns='Temperature category', values='Total_count', aggfunc='sum', fill_value=0)

print("Группировка через groupby:")
display(group)

print("Сводная таблица через pivot_table:")
display(pivot)
```

10. Найти сезон и категорию с максимумом и минимумом аренд

```python
max_row = group.loc[group['Total_count'].idxmax()]
min_row = group.loc[group['Total_count'].idxmin()]

print("Максимум аренд:")
print(max_row)

print("Минимум аренд:")
print(min_row)
```
