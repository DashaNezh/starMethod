импорт таблицы через url, берем первую таблицу [0]

```python
import pandas as pd
```

```python
df = pd.read_html("https://pstu.ru/files/file/Abitur/2025%20final/%D0%A4%D0%9F%D0%9C%D0%9C_%D0%91_%D0%9D_000007052.html")[0]
df
```

убираем лишние строки сверху

```python
df = df.iloc[13:]
df
```

убираем пустые столбцы (axis=1 - работаем по столбцам, how="all" - удаляем только те столбцы, где все пустые значения)

```python
df = df.dropna(axis=1, how="all")
df
```

переименовываем столбцы

```python
df.columns = ["№",	"Уникальный код",	"Сумма баллов",	"АиНМА / Математика",	"ИТ / Информатика и ИКТ / Физика",	"Русский язык",	"Сумма баллов за ИД",	"Учебная группа",	"Представление приказа"]
df
```

убираем не нужные столбцы

```python
df = df.drop(columns=["№",	"Уникальный код", "Учебная группа",	"Представление приказа"])
df
```

переименовать порядковые индексы (drop=True - убирает сохранение старого индекса в отдельном столбце)

```python
df = df.reset_index(drop=True)
df
```

Проверяем типы данных во всех столбцах

```python
print(df.dtypes)
```

приводим к числовому типу (astype - можно если все значения числа)

```python
df = df.astype(int)
print(df.dtypes)
```

Определить средний, минимальный и максимальный суммарные баллы.

```python
mean_value = df["Сумма баллов"].mean()
min_value = df["Сумма баллов"].min()
max_value = df["Сумма баллов"].max()

print("Средний: ", mean_value)
print("Минимальный: ", min_value)
print("Максимальный: ", max_value)
```

Определить средние значения по трем сдаваемым предметам ЕГЭ

```python
data = ["АиНМА / Математика",	"ИТ / Информатика и ИКТ / Физика",	"Русский язык"]
mean_value = df[data].mean()

mean_df = pd.DataFrame(mean_value, columns=["Средний балл"])

mean_df
```

предмет, по которому абитуриенты в среднем набирают больше баллов.

```python
max_mean = mean_df["Средний балл"].max()     # максимальный балл
max_subject = mean_df["Средний балл"].idxmax()    # индекс строки с максимальным баллом 

print("Предмет с максимальным баллом: ", max_subject, max_mean)
```

количество абитуриентов, у которых балла ЕГЭ по русскому ниже
среднего балла по русскому языку, а по остальным предметам выше средних

```python
mean_math = df["АиНМА / Математика"].mean()
mean_it = df["ИТ / Информатика и ИКТ / Физика"].mean()
mean_russian = df["Русский язык"].mean()

# выборка подходящих значений
type_1 = df[
    (df["Русский язык"]<mean_russian)&(df["АиНМА / Математика"]>mean_math) & (df["ИТ / Информатика и ИКТ / Физика"]>mean_it)]

# кол-во строк подходящих под условия
count_1=len(type_1)

print("количество абитуриентов (тип 1): ", count_1)
```

количество абитуриентов, у которых балла ЕГЭ по русскому выше
среднего балла по русскому языку, а по остальным предметам ниже средних

```python
# выборка подходящих значений
type_2 = df[
    (df["Русский язык"]>mean_russian)&(df["АиНМА / Математика"]<mean_math) & (df["ИТ / Информатика и ИКТ / Физика"]<mean_it)]

# кол-во строк подходящих под условия
count_2=len(type_2)

print("количество абитуриентов (тип 2): ", count_2)
```

итоговая таблица

```python
mean_value_1 = df["Сумма баллов"].mean()

new_df = pd.DataFrame({
    "Подразделение": ["Факультет прикладной математики и механики"],
    "Уровень подготовки": ["Бакалавриат"],
    "Направление подготовки/специальность": ["Информационные системы и технологии"],
    "Год": ["2025"],
    "Количество мест": [len(df)],
    "Предметы ЕГЭ": ["АиНМА / Математика, ИТ / Информатика и ИКТ / Физика, Русский язык"],
    "Средняя сумма баллов": [mean_value_1],
    "Min сумма баллов": [min_value],
    "Max сумма баллов": [max_value],
    "Предмет с высшим средним": [max_subject],
    "Кол-во студентов тип_1": [count_1],
    "Кол-во студентов тип_2": [count_2],
})

# транспонируем таблицу для читаемого вывода
new_df.transpose()
```
