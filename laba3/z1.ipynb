```python
import pandas as pd
```

```python
ratings = pd.read_csv("ml-1m/ratings.dat", sep="::", names=["UserID", "MovieID", "Rating", "Timestamp"], engine="python")
users = pd.read_csv("ml-1m/users.dat", sep="::", names=["UserID", "Gender", "Age", "Occupation", "Zip-code"], engine="python")
movies = pd.read_csv("ml-1m/movies.dat", sep="::", names=["MovieID", "Title", "Genres"], engine="python", encoding="latin-1")

# pd.merge - объединяет по одному/нескольким столбцам или индексам
ratings_users = pd.merge(ratings, users, on="UserID") # по столбцу UserID
general_table = pd.merge(ratings_users, movies, on="MovieID") # по столбцу MovieID
general_table
```

1. Требуется вычислить средние рейтинги для фильма с ID = № варианта в зависимости
от пола; 44

```python
female=general_table[(general_table["MovieID"]==44) & (general_table["Gender"]=="F")] # булевая фильтрация тру или фолс
print("Средний рейтинг для фильма с ID = 44 для женщин: ", female["Rating"].mean())

man = general_table[(general_table["MovieID"]==44) & (general_table["Gender"]=="M")]
print("Средний рейтинг для фильма с ID = 44 для мужчин: ", man["Rating"].mean())
```

2. Отобрать фильмы, получившие не менее 250 оценок, и для них определить средний
рейтинг в зависимости от пола;
   - count() - подсчитывает количество непустых (не-NaN) значений в каждой колонке или строке

```python
id = general_table.groupby("MovieID")["Rating"].count().reset_index(name="Count Rating") #кол-во оценок для каждого фильма
id
```

объединение таблиц

```python
general_table = pd.merge(general_table, id, on="MovieID")
general_table
```

фильмы с кол-вом оценок более 250

```python
count_rating_250 = general_table[general_table["Count Rating"] >= 250]
count_rating_250
```

```python
average_rating = count_rating_250.groupby(["MovieID", "Title", "Gender"])["Rating"].mean().reset_index()
average_rating
```

```python
rating = average_rating.pivot_table(index=["MovieID", "Title"], columns="Gender", values="Rating")
rating.columns = ["Rating_F", "Rating_M"]
rating.reset_index()
```

3. Найти фильмы, оказавшиеся в топ 5 у зрителей-женщин и худшие 5 у зрителей-
мужчин. Найти фильмы (топ 5), по которым мужчины и женщины сильнее всего
разошлись в оценках;

```python
top_rating = general_table.pivot_table(index="Title", columns="Gender", values="Rating", aggfunc="mean").reset_index()
top_rating
```

топ 5 у зрителей-женщин

```python
female_top_5 = top_rating.sort_values(by="F", ascending=False)
female_top_5["Title"].iloc[:5].reset_index()
```

худшие 5 у зрителей-мужчин

```python
man_worst_5 = top_rating.sort_values(by="M", ascending=True)
man_worst_5["Title"].iloc[:5].reset_index()
```

Найти фильмы (топ 5), по которым мужчины и женщины сильнее всего разошлись в оценках

```python
top_rating["F"] = top_rating["F"].fillna(0) # заменяем NAN на 0
top_rating["M"] = top_rating["M"].fillna(0) # заменяем NAN на 0

top_rating["diff"] = abs(top_rating["F"] - top_rating["M"])

top_rating = top_rating.sort_values(by="diff", ascending=False)
top_rating[["Title", "F", "M","diff"]].iloc[:5].reset_index()
```

4. Определить фильмы (топ 5), вызвавшие в среднем наибольшее разногласие у зрителей независимо от пола;

Стандартное отклонение (std), показывает разброс данных относительно среднего значения

```python
std_rating = general_table.groupby("Title")["Rating"].std().reset_index(name="std_rating")

std_rating_5 = std_rating.sort_values(by="std_rating", ascending=False)
std_rating_5.iloc[:5].reset_index()
```

5. Определить средний рейтинг каждого жанра для различных возрастов. Установите,
фильмы какого жанра в каждой категории возрастов имеют наибольший и наименьший
рейтинги.
   - str - применяем split к каждой строке текущего столбца

```python
general_table["Genres"] = general_table["Genres"].str.split("|")
general_table = general_table.explode("Genres") # разбиваем на строки для каждого жарна
general_table
```

```python
general_table_genres = general_table.groupby(["Age", "Genres"])["Rating"].mean().reset_index()
general_table_genres
```

наибольший рейтинг

```python
age_table_max = general_table_genres.groupby("Age")["Rating"].max().reset_index()

max_rating = pd.merge(general_table_genres, age_table_max, on=["Age", "Rating"])
max_rating
```

наименьший рейтинг

```python
age_table_min = general_table_genres.groupby("Age")["Rating"].min().reset_index()

min_rating = pd.merge(general_table_genres, age_table_min, on=["Age", "Rating"])
min_rating
```
